```python
class Account:
    def __init__(self):
        self._balance = 0

    @property
    def balance(self):
        return self._balance

    def deposit(self, n):
        self._balance += n

    def withdraw(self, n):
        self._balance -= n


```

```python
account = Account()
print(account.balance)
account.deposit(400)
account.withdraw(299)
print(account.balance)
```

```python
class Solution:
    def smalestNum(self, nums):
        smallest_num = min(nums)
        return smallest_num

nums = [4, 5, 8, 2, 9, 1]
solution = Solution()
smallest_num = solution.smalestNum(nums)
print(smallest_num)
```

```python
class Solution:
    def secondLargest(self, nums, n):
        large_num = nums[0]
        second_large = nums[0]
        for i in range(1, n):
            if nums[i] > large_num:
                large_num = nums[i]
            elif nums[i] > second_large and nums[i] != large_num:
                second_large = nums[i]
        return second_large

nums = [2, 5, 4 ,3, 9]
n = len(nums)
solution =  Solution()
second_large_num = solution.secondLargest(nums, n)
print(second_large_num)
```

```python
class Quantity:

    def __init__(self, food_id, m_q=0, a_q=0, d_q=0, r_q=0, g_q=0, p_q=0, b_q=0):
        if m_q == m_q:
            food_id = 1
        elif a_q == a_q:
            food_id = 1
        self.food_id = food_id
        self.m_q = m_q
        self.a_q = a_q
        self.d_q = d_q
        self.r_q = r_q
        self.g_q = g_q
        self.p_q = p_q
        self.b_q = b_q

    def __str__(self):
        return f"Hello{self.food_id}"
```

```python
def main():
    food_id = input()
    quantity = Quantity(food_id)
    print(quantity)
```

```python
if __name__ == "__main__":
    main()
```
